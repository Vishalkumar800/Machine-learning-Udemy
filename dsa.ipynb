```python
# Store frequency in dictionary mtlb kitni baar aaya hai.

a = [2,3,5,2,3,5,6,6,3,2,2,21,1,1,1,1]

dict = {}

for i in a:
    if i in dict:
        dict[i] +=1
    else:
        dict[i] = 1

print(dict) #   TC = O(n)
# Space Complexity O(n) # dict 
```
