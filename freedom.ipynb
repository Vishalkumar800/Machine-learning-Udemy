Definition:

The symbol used for summation is Σ (Sigma).

General Form:
$$ \sum_{i=1}^{n} i $$


“This means adding all the values of \(i\) from 1 to \(n\).”


i=1+2+3+⋯+n

## Measure of Central Tendancy

Measure of Central Tendency is a statistical measure that describes the central or typical value of a dataset. It mainly includes Mean, Median, and Mode.

```

Mean: The average value of the data.
Mode: The value that occurs most frequently in a dataset.
Median: The middle value of the data when the data is arranged in ascending or descending order.


Question : 

2, 2, 2, 3, 4, 4, 5, 5, 6, 8, 8, 8, 9, 9, 11, 23, 28, 28, 34, 36, 42 , 8 , 8 

```

````python
import numpy as np 

data = [2, 2, 2, 3, 4, 4, 5, 5, 6, 8, 8, 8, 9, 9, 11, 23, 28, 28, 34, 36, 42 , 8 , 8 ]

mean = np.mean(data)
median = np.median(data)

print(mean)
print(median)


````

### When Do We Use Mean, Median and Mode?

**1. Mean – When the data is balanced**

We use **Mean** when the data is relatively balanced and does not have extreme values (outliers).

**Example:**
`10, 12, 13, 14, 15`

Here, the values are close to each other, so the **Mean** is a good measure of the center.

---

**2. Median – When the data has outliers or is spread/skewed**

We use **Median** when some values are very far away from the other values.

**Example:**
`10, 12, 13, 14, 100`

Here, `100` is an outlier. It can make the Mean misleading, so we prefer the **Median**.

---

**3. Mode – When repetition/frequency is important**

We use **Mode** when we want to know which value occurs **most frequently**.

**Example:**
`2, 3, 3, 3, 5, 7`

Here, `3` occurs most frequently, so the **Mode = 3**.

Mode is also very useful for **categorical data**.

**Example:**
`Red, Blue, Red, Green, Red`

Here, **Mode = Red** because Red occurs most frequently.

### Easy Rule to Remember

* **Mean → Balanced data**
* **Median → Outliers / Skewed data**
* **Mode → Most frequent value**
