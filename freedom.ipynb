## Summation (Σ)

Definition:
Summation means adding a series of numbers together.

The symbol used for summation is Σ (Sigma).

General Form:
$$ \sum_{i=1}^{n} i $$


“This means adding all the values of \(i\) from 1 to \(n\).”


i=1+2+3+⋯+n
